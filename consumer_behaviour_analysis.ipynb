```python
import pandas as pd
df = pd.read_csv("customer_shopping_behavior.csv")
df.head(10)
```

```python
df.isna().sum()
```

```python
df.columns = df.columns.str.lower().str.replace(" ", "_")
df.columns
```

```python
df[df["review_rating"].isna()]
```

```python
df["review_rating"] = df.groupby("category")["review_rating"].transform(lambda x: x.fillna(x.median()))
df["review_rating"].isna()
```

```python
df = df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
df.columns
```

```python
df.info()
```

```python
df['age_group'] = pd.cut(df['age'], bins=[0, 25, 40, 60, 100],labels=["Young Adult", "Adult", "Middle Aged", "Senior"])
df[['age','age_group']]
```

```python
df["purchase_fequency_days"] = df ["frequency_of_purchases"].map({'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90})
df[["purchase_fequency_days","frequency_of_purchases"]]
```

```python
df.columns
```

```python
df[['discount_applied','promo_code_used']].head(10)
```

```python
df = df.drop(columns='promo_code_used')
```
